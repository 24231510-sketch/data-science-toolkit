```python
#Part 1-Google Colab and Python
# 1. Create a new notebook and code the following
# 1. Create a numpy array array of size 2 x 2

#numpy is a python library that allows you to work with numbers and large datasets etc
import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("Array 2x2")
print(array_2x2)
```

```python
# 2. Create a random numpy array of size 3 x 4

#The first number is the row and second is the column
#So this is a grid of 3 rows and 4 columns

#this allows you to make an array of shape 3,4 and fill it with random numbers. Your numbers will be float
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeroes

ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones:")
print(ones_array)

print("Array of zeros:")
print(zeros_array)
```

```python
#4.Create a random numpy array of a size 6x3 and transpose it.
random6x3 = np.random.rand(6,3)
transposed_random6x3 = np.transpose(random6x3)

print("Array of random 6x3:")
print(random6x3)

print("Transposed array of random 6x3")
print(transposed_random6x3)
```

```python
#5. Create two random number array of size n x p and p x m and then compute the dot product
n=3
p=4
m=5
Array1 = np.random.rand(n,p)
Array2 = np.random.rand(p,m)

dot_product = np.dot(Array1,Array2)

print("Array 1:")
print(Array1)

print("Array 2:")
print(Array2)

print("Dot product:")
print(dot_product)
```
